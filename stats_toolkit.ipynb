```python
def read_sales(path):
    with open(path, "r", newline="") as file:
        reader = csv.DictReader(file)
        rows = list(reader)
        print(f"Rows: {len(rows)}")
        print(f"Columns: {reader.fieldnames}")
        return rows
        

read_sales(r"C:\Users\hp\desktop\stats_toolkit\data\agent_deposits.csv") 
```

```python
deposits = []
rejected = []

with open(DATA_PATH, "r", newline="") as file:

    reader = csv.DictReader(file)

    for row_number, row in enumerate(reader, start=2):

        raw_value = row["daily_deposits_kes"]

        try:
            value = float(raw_value)
            deposits.append(value)

        except ValueError:
            rejected.append((row_number, raw_value))

print("Valid deposits:", deposits)
print("Rejected values:", rejected)
```

```python
def mean(values):
    if len(values) == 0:
        raise ValueError

    return sum(values) / len(values)
mean(deposits)
```

```python
def median(values):
    if not values:
        raise ValueError

    sorted_values = sorted(values)
    n = len(sorted_values)
    middle = n // 2

    if n % 2 == 1:
        return sorted_values[middle]
    else:
        return (sorted_values[middle - 1] + sorted_values[middle]) / 2

median(deposits)
```

```python
def mode(values):
    if not values:
        raise ValueError

    counts = {}

    for value in values:
        counts[value] = counts.get(value, 0) + 1

    highest_count = max(counts.values())

    if highest_count == 1:
        return []

    return [
        value for value, count in counts.items()
        if count == highest_count
    ]
mode(deposits)
```

Population Variance 

```python
def variance(values, sample=False):
    if not values:
        raise ValueError

    if sample and len(values) < 2:
        raise ValueError

    average = mean(values)

    squared_differences = []

    for value in values:
        difference = value - average
        squared = difference ** 2
        squared_differences.append(squared)

    if sample:
        return sum(squared_differences) / (len(values) - 1)

    return sum(squared_differences) / len(values)
variance(deposits, sample=False)
```

Sample Variance

```python
def variance(values, sample=True):
    if not values:
        raise ValueError

    if sample and len(values) < 2:
        raise ValueError

    average = mean(values)

    squared_differences = []

    for value in values:
        difference = value - average
        squared = difference ** 2
        squared_differences.append(squared)

    if sample:
        return sum(squared_differences) / (len(values) - 1)

    return sum(squared_differences) / len(values)
variance(deposits, sample=True)
```

```python
def std_dev(values, sample=False):
    return math.sqrt(variance(values, sample))
std_dev(deposits, sample=False)
```

```python
def quartiles(values):
    if not values:
        raise ValueError("Cannot calculate quartiles of an empty list")

    sorted_values = sorted(values)
    n = len(sorted_values)

    middle = n // 2

    if n % 2 == 0:
        lower_half = sorted_values[:middle]
        upper_half = sorted_values[middle:]
    else:
        lower_half = sorted_values[:middle]
        upper_half = sorted_values[middle + 1:]

    q1 = median(lower_half)
    q2 = median(sorted_values)
    q3 = median(upper_half)

    return q1, q2, q3
quartiles(deposits)
```

```python
def iqr_outliers(values, k=1.5):
    q1, q2, q3 = quartiles(values)

    iqr = q3 - q1

    lower_fence = q1 - (k * iqr)
    upper_fence = q3 + (k * iqr)

    outliers = []

    for value in values:
        if value < lower_fence or value > upper_fence:
            outliers.append(value)

    return iqr, lower_fence, upper_fence, outliers
iqr_outliers(deposits)
```

```python
class StatisticsAnalyzer:

    def __init__(self, values, rejected=None, sample=False, k=1.5):
        self.values = values
        self.rejected = rejected if rejected is not None else []
        self.sample = sample
        self.k = k

       
        self.sorted_values = sorted(values)

        
        self._summary = self._calculate_summary()

    def _calculate_summary(self):
        q1, q2, q3 = quartiles(self.sorted_values)

        iqr, lower, upper, outliers = iqr_outliers(
            self.sorted_values, self.k
        )

        return {
            "count": len(self.values),
            "mean": mean(self.values),
            "median": median(self.sorted_values),
            "mode": mode(self.values),
            "variance": variance(self.values, self.sample),
            "standard_deviation": std_dev(self.values, self.sample),
            "q1": q1,
            "q2": q2,
            "q3": q3,
            "iqr": iqr,
            "lower_fence": lower,
            "upper_fence": upper,
            "outliers": outliers
        }

    @classmethod
    def from_csv(cls, path, column, sample=False, k=1.5):
        values, rejected = load_column(path, column)

        return cls(values, rejected, sample, k)

    def summary(self):
        return self._summary.copy()

    def without_outliers(self):
        lower = self._summary["lower_fence"]
        upper = self._summary["upper_fence"]

        filtered_values = [
            value for value in self.values
            if lower <= value <= upper
        ]

        return StatisticsAnalyzer(
            filtered_values,
            self.rejected,
            self.sample,
            self.k
        )

    def report(self):
        results = self.summary()

        report_text = f"""
WAKALA PLUS - DAILY DEPOSIT STATISTICS

Number of records: {results['count']}
Mean: {results['mean']}
Median: {results['median']}
Mode: {results['mode']}

Variance: {results['variance']}
Standard Deviation: {results['standard_deviation']}

Q1: {results['q1']}
Q2: {results['q2']}
Q3: {results['q3']}
IQR: {results['iqr']}

Lower Fence: {results['lower_fence']}
Upper Fence: {results['upper_fence']}
Outliers: {results['outliers']}

Rejected Records: {len(self.rejected)}
"""
        return report_text
    
analyzer = StatisticsAnalyzer(deposits)
```
